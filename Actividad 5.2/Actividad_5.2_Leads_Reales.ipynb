**PREPROCESAMIENTO - VALORES ATÍPICOS (Rango Intercuartílico)**

```python
#Creo 2 dataframe para poder procesar los outliers
cualitativas= data3[['Año','Mes']]
cuantitativas= data3[['Total','Ilocalizable','No Contesta','D. Incorrectos','Duplicados','Efectivos','Abiertos','Ventas','Conversion ','Efectividad de Leads']]
```

```python
#Realizamos diagrama de caja de cada columna del dataframe
fig = plt.figure(figsize =(15, 8))
cuantitativas.plot(kind='box', vert=False)
plt.title("Valores Atípicos - datos originales")
plt.show()
```

```python
#Método aplicando Cuartiles. Encuentro cuartiles 0.25 y 0.75
y=cuantitativas

percentile25=y.quantile(0.25) #Q1
percentile75=y.quantile(0.75) #Q3
iqr= percentile75 - percentile25

Limite_Superior_iqr= percentile75 + 1.5*iqr
Limite_Inferior_iqr= percentile25 - 1.5*iqr
print("Limite superior permitido", Limite_Superior_iqr)
print("Limite inferior permitido", Limite_Inferior_iqr)
```

```python
#Obtenemos datos limpios del Dataframe
data_iqr= cuantitativas[(y<=Limite_Superior_iqr)&(y>=Limite_Inferior_iqr)]
data_iqr
```

```python
#Corroboramos valores nulos del dataframe
valores_nulos=data_iqr.isnull().sum()
valores_nulos
```

```python
#Reemplazamos valores atípicos (nulos) del dataframe con "median"
data4_iqr=data_iqr.copy()
data4_iqr=data4_iqr.fillna(round(data_iqr.median(),1))
data4_iqr
```

```python
#Corroboramos valores nulos del dataframe LIMPIO
valores_nulos=data4_iqr.isnull().sum()
valores_nulos
```

```python
# Unimos el dataframe cuantitativo limpio con el dataframe cualitativo
Datos_limpios = pd.concat([cualitativas, data4_iqr], axis=1)
Datos_limpios
```

**TRATAR COLUMNAS PARA QUE RESULTEN DE TIPO NUMÉRICO**

```python
#Reemplazamos los valores categoricos de Mes con el numero de mes correspondiente
meses = {'Enero':1,'Febrero':2,'Marzo':3,'Abril':4,'Mayo':5,'Junio':6,
         'Julio':7,'Agosto':8,'Septiembre':9,'Octubre':10,'Noviembre':11,'Diciembre':12}
Datos_limpios['Mes'] = Datos_limpios['Mes'].str.strip().replace(meses).astype(int)
Datos_limpios['Año'] = Datos_limpios['Año'].astype(int)
Datos_limpios.dtypes
```

**ANÁLISIS DE CORRELACIÓN**

```python
#Columnas de tipo de Leads a analizar
leads_cols = ['Total','Ilocalizable','No Contesta','D. Incorrectos','Duplicados','Efectivos','Abiertos','Ventas']
```

```python
#Graficamos las dispersiones entre las variables de tipo de Leads
sns.pairplot(Datos_limpios[leads_cols])
```

```python
#Encontramos todas las correlaciones entre las variables de tipo de Leads
Corr_Factors = Datos_limpios[leads_cols].corr()
Corr_Factors
```

```python
#Encontramos el valor absoluto de todas las correlaciones
Corr_Factors1 = abs(Corr_Factors)
Corr_Factors1
```

```python
#Graficamos el mapa de calor de los coeficientes de correlación
fig = plt.figure(figsize=(9,7))
Heat_Map = sns.heatmap(Corr_Factors1, cmap='Greens', annot=True, fmt=".2f")
Heat_Map
```

**Efectivos es la variable con mayor correlación respecto a Total, así que la usamos como variable independiente para la Regresión Lineal Simple**

```python
#Imprimimos el scatter plot entre la variable dependiente (Total) e independiente (Efectivos)
sns.scatterplot(x='Efectivos', y='Total', color="blue", data=Datos_limpios)
```

```python
#Declaramos las variables dependiente e independiente para la regresión lineal
Vars_Indep = Datos_limpios[['Efectivos']]
Var_Dep = Datos_limpios['Total']
```

```python
#Se define model como la función de regresión lineal
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
#Ajustamos el modelo con las variables antes declaradas
model.fit(X=Vars_Indep, y=Var_Dep)
```

```python
#Verificamos los coeficientes obtenidos para el modelo ajustado
model.__dict__
```

**Modelo matemático: Total = coef_ · Efectivos + intercept_**

```python
#Evaluamos la eficiencia del modelo obtenido por medio del coeficiente R Determinación
model.score(Vars_Indep, Var_Dep)
```

```python
#Predecimos los valores de Total a partir de la variable Efectivos
y_pred = model.predict(X=Datos_limpios[['Efectivos']])
y_pred
```

```python
#Insertamos la columna de predicciones en el DataFrame
Datos_limpios.insert(0, 'Predicciones', y_pred)
Datos_limpios
```

```python
#Visualizamos la gráfica comparativa entre el Total real y el Total predecido
sns.scatterplot(x='Efectivos', y='Total', color="blue", data=Datos_limpios)
sns.scatterplot(x='Efectivos', y='Predicciones', color="red", data=Datos_limpios)
sns.lineplot(x='Efectivos', y='Predicciones', color="red", data=Datos_limpios)
```

```python
#Corroboramos cual es el coeficiente de Determinación de nuestro modelo
coef_Deter = model.score(X=Vars_Indep, y=Var_Dep)
coef_Deter
```

```python
#Corroboramos cual es el coeficiente de Correlación de nuestro modelo
coef_Correl = np.sqrt(coef_Deter)
coef_Correl
```
