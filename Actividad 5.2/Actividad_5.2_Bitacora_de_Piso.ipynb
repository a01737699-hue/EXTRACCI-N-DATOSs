**PREPROCESAMIENTO - VALORES ATÍPICOS**

PDM, SDC, Venta e Inter Gerente son variables booleanas (0/1), no continuas, por lo que no aplica el procedimiento de desviación estándar / cuartiles para outliers.

**TRATAR COLUMNAS PARA QUE RESULTEN DE TIPO NUMÉRICO**

```python
#Convertimos las variables booleanas a tipo numerico (0/1)
for col in ['PDM','SDC','Venta','Inter Gerente']:
    data3[col] = data3[col].astype(int)
data3[['PDM','SDC','Venta','Inter Gerente']].dtypes
```

**ANÁLISIS DE CORRELACIÓN**

```python
#Columnas numericas a analizar
numericas = data3[['PDM','SDC','Venta','Inter Gerente']]
```

```python
#Graficamos las dispersiones entre las variables
sns.pairplot(numericas)
```

```python
#Encontramos todas las correlaciones entre las variables
Corr_Factors = numericas.corr()
Corr_Factors
```

```python
#Encontramos el valor absoluto de todas las correlaciones
Corr_Factors1 = abs(Corr_Factors)
Corr_Factors1
```

```python
#Graficamos el mapa de calor de los coeficientes de correlación
fig = plt.figure(figsize=(7,5))
Heat_Map = sns.heatmap(Corr_Factors1, cmap='Greens', annot=True, fmt=".2f")
Heat_Map
```

**Inter Gerente es la variable con mayor correlación respecto a Venta (0.24), así que la usamos como variable independiente para la Regresión Lineal Simple**

```python
#Imprimimos el scatter plot entre la variable dependiente (Venta) e independiente (Inter Gerente)
sns.scatterplot(x='Inter Gerente', y='Venta', color="blue", data=data3)
```

```python
#Declaramos las variables dependiente e independiente para la regresión lineal
Vars_Indep = data3[['Inter Gerente']]
Var_Dep = data3['Venta']
```

```python
#Se define model como la función de regresión lineal
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
#Ajustamos el modelo con las variables antes declaradas
model.fit(X=Vars_Indep, y=Var_Dep)
```

```python
#Verificamos los coeficientes obtenidos para el modelo ajustado
model.__dict__
```

**Modelo matemático: Venta = coef_ · Inter Gerente + intercept_**

```python
#Evaluamos la eficiencia del modelo obtenido por medio del coeficiente R Determinación
model.score(Vars_Indep, Var_Dep)
```

```python
#Predecimos los valores de Venta a partir de la variable Inter Gerente
y_pred = model.predict(X=data3[['Inter Gerente']])
y_pred
```

```python
#Insertamos la columna de predicciones en el DataFrame
data3.insert(0, 'Predicciones', y_pred)
data3
```

```python
#Visualizamos la gráfica comparativa entre Venta real y Venta predecida
sns.scatterplot(x='Inter Gerente', y='Venta', color="blue", data=data3)
sns.scatterplot(x='Inter Gerente', y='Predicciones', color="red", data=data3)
sns.lineplot(x='Inter Gerente', y='Predicciones', color="red", data=data3)
```

```python
#Corroboramos cual es el coeficiente de Determinación de nuestro modelo
coef_Deter = model.score(X=Vars_Indep, y=Var_Dep)
coef_Deter
```

```python
#Corroboramos cual es el coeficiente de Correlación de nuestro modelo
coef_Correl = np.sqrt(coef_Deter)
coef_Correl
```
