```python
#Variables que posteriormente serán categorizadas con la regla de Sturges
variables_leads = ['Total', 'Ilocalizable', 'No Contesta', 'D. Incorrectos',
                   'Duplicados', 'Efectivos', 'Abiertos', 'Ventas',
                   'Conversion', 'Efectividad de Leads']

#Eliminamos comas, signos de porcentaje y convertimos las variables a numéricas
for variable in variables_leads:
    Leads_Reales[variable] = (
        Leads_Reales[variable]
        .astype(str)
        .str.replace(',', '', regex=False)
        .str.replace('%', '', regex=False)
        .str.strip()
    )

    Leads_Reales[variable] = pd.to_numeric(
        Leads_Reales[variable],
        errors='coerce'
    )
```

```python
# ANÁLISIS UNIVARIADO DE 10 VARIABLES CATEGÓRICAS
# Base de datos: Bitácora de Piso

# Convertimos Fecha a formato de fecha
Bitacora_Piso['Fecha'] = pd.to_datetime(
    Bitacora_Piso['Fecha'],
    errors='coerce'
)

# Creamos dos variables categóricas a partir de Fecha
Bitacora_Piso['Mes'] = Bitacora_Piso['Fecha'].dt.month_name()

Bitacora_Piso['Dia_semana'] = Bitacora_Piso['Fecha'].dt.day_name()

# Cambiamos los nombres al español
meses_espanol = {
    'January':'Enero',
    'February':'Febrero',
    'March':'Marzo',
    'April':'Abril',
    'May':'Mayo',
    'June':'Junio',
    'July':'Julio',
    'August':'Agosto',
    'September':'Septiembre',
    'October':'Octubre',
    'November':'Noviembre',
    'December':'Diciembre'
}

dias_espanol = {
    'Monday':'Lunes',
    'Tuesday':'Martes',
    'Wednesday':'Miércoles',
    'Thursday':'Jueves',
    'Friday':'Viernes',
    'Saturday':'Sábado',
    'Sunday':'Domingo'
}

Bitacora_Piso['Mes'] = Bitacora_Piso['Mes'].replace(meses_espanol)
Bitacora_Piso['Dia_semana'] = Bitacora_Piso['Dia_semana'].replace(dias_espanol)


# Seleccionamos las 10 variables categóricas
variables_categoricas = [
    'Asesor',
    'Esatus Lead',
    'PDM',
    'SDC',
    'Estatus de SDC',
    'Venta',
    'Temperatura',
    'Inter Gerente',
    'Mes',
    'Dia_semana'
]


# Creamos espacio para las 10 gráficas
fig, axes = plt.subplots(5, 2, figsize=(16, 22))

axes = axes.flatten()

# Diccionario para guardar tablas
tablas_frecuencia = {}


# Realizamos el análisis univariado
for i, variable in enumerate(variables_categoricas):

    # Tabla de frecuencias
    Tabla_freq = Bitacora_Piso[variable].value_counts().reset_index()

    # Guardamos tabla
    tablas_frecuencia[variable] = Tabla_freq

    # Ajustamos índice
    Tabla_index = Tabla_freq.set_index(variable)

    # Gráfica
    Tabla_index.plot(
        kind='bar',
        ax=axes[i],
        width=0.8,
        legend=False
    )

    axes[i].set_title(variable)
    axes[i].set_xlabel('')
    axes[i].set_ylabel('Frecuencia')
    axes[i].tick_params(axis='x', rotation=45)


plt.tight_layout()
plt.show()
```

```python
# CATEGORIZACIÓN DE 10 VARIABLES NUMÉRICAS
# Base de datos: Leads Reales
# Aplicando la regla de Sturges

variables_leads = [
    'Total',
    'Ilocalizable',
    'No Contesta',
    'D. Incorrectos',
    'Duplicados',
    'Efectivos',
    'Abiertos',
    'Ventas',
    'Conversion',
    'Efectividad de Leads'
]

# Diccionarios para guardar resultados
tablas_sturges = {}
resumen_sturges = {}

# Creamos espacio para las 10 gráficas
fig, axes = plt.subplots(5, 2, figsize=(16, 22))
axes = axes.flatten()


# Recorremos las 10 variables
for j, variable in enumerate(variables_leads):

    # Número de datos
    n = Leads_Reales[variable].count()

    # Obtenemos limite superior e inferior
    Max = Leads_Reales[variable].max()
    Min = Leads_Reales[variable].min()

    # Calculamos el rango R
    R = Max - Min

    # Regla de Sturges
    ni = 1 + 3.32 * np.log10(n)

    # Redondeamos hacia arriba el número de intervalos
    ni_entero = int(np.ceil(ni))

    # Calculamos el ancho del intervalo
    i = R / ni_entero

    # Creamos los intervalos
    intervalos = np.linspace(Min, Max, ni_entero + 1)

    # Creamos nombres para las categorías
    categorias = []

    for x in range(ni_entero):
        categorias.append(
            f'Cat{x+1} ({intervalos[x]:.2f}-{intervalos[x+1]:.2f})'
        )

    # Categorizamos la variable
    Leads_Reales[variable + '_Categoria'] = pd.cut(
        Leads_Reales[variable],
        bins=intervalos,
        labels=categorias,
        include_lowest=True
    )

    # Tabla de frecuencias de las categorías
    Tabla_freq = (
        Leads_Reales[variable + '_Categoria']
        .value_counts(sort=False)
        .reset_index()
    )

    # Guardamos la tabla
    tablas_sturges[variable] = Tabla_freq

    # Guardamos los resultados de Sturges
    resumen_sturges[variable] = {
        'n': n,
        'Min': Min,
        'Max': Max,
        'Rango': R,
        'Sturges': ni,
        'Intervalos': ni_entero,
        'Ancho': i
    }

    # Ajustamos el índice para graficar
    Tabla_index = Tabla_freq.set_index(variable + '_Categoria')

    # Gráfica de barras
    Tabla_index.plot(
        kind='bar',
        ax=axes[j],
        width=0.8,
        legend=False
    )

    axes[j].set_title(variable)
    axes[j].set_xlabel('Categorías')
    axes[j].set_ylabel('Frecuencia')
    axes[j].tick_params(axis='x', rotation=45)


# Ajustamos las gráficas
plt.tight_layout()

# Mostramos las 10 gráficas juntas
plt.show()
```

```python
#Mostramos los cálculos realizados con la regla de Sturges

Resumen_Sturges = pd.DataFrame(resumen_sturges).T

Resumen_Sturges
```

```python
# Ensure 'PDM' is numeric for calculation (True=1, False=0)
Bitacora_Piso['PDM_numeric'] = Bitacora_Piso['PDM'].astype(int)

# Group by 'Asesor' and aggregate 'PDM_numeric' to create ct_asesor_piso
ct_asesor_piso = Bitacora_Piso.groupby('Asesor')['PDM_numeric'].agg([
    ('Total', 'count'), # Total leads attended by each asesor
    ('Tasa PDM (%)', lambda x: x.mean() * 100) # Percentage of PDM for each asesor
])

# Monstrar el top 15
data_plot = ct_asesor_piso.sort_values('Total', ascending=False).head(15).sort_values('Total', ascending=True)

fig, ax1 = plt.subplots(figsize=(9,5.5))
ax1.barh(data_plot.index, data_plot['Total'], color="#4c72b0", alpha=0.6, label='Leads atendidos')
ax1.set_xlabel('Leads atendidos')
ax1.set_title('Volumen vs. tasa de prueba de manejo por asesor (Top 15)')

ax2 = ax1.twiny()
ax2.plot(data_plot['Tasa PDM (%)'], data_plot.index, 'o-', color="#c44e52", linewidth=2, markersize=7, label='Tasa PDM')
ax2.set_xlabel('Tasa de prueba de manejo (%)')

lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1+lines2, labels1+labels2, loc='lower right')
plt.tight_layout()
plt.show()
```

```python
mes_map={'Enero':1,'Febrero':2,'Marzo':3,'Abril':4,'Mayo':5,'Junio':6,
          'Julio':7,'Agosto':8,'Septiembre':9,'Octubre':10,'Noviembre':11,'Diciembre':12}
Leads_Reales['periodo'] = [pd.Period(year=int(a), month=mes_map[m], freq='M')
                     for a, m in zip(Leads_Reales['Año'], Leads_Reales['Mes'])]
Leads_Reales = Leads_Reales.sort_values('periodo')

motivos = ['Ilocalizable','No Contesta','D. Incorrectos','Duplicados']
for c in motivos:
    Leads_Reales[c+'_pct'] = Leads_Reales[c] / Leads_Reales['Total'] * 100

fig, ax = plt.subplots(figsize=(11,5.5))
ax.stackplot(Leads_Reales['periodo'].astype(str), [Leads_Reales[c+'_pct'] for c in motivos], labels=motivos,
             colors=['#c44e52','#dd8452','#937860','#8172b2'], alpha=0.85)
ax.set_ylabel('% de leads del mes')
ax.set_title('Motivo de pérdida de leads a lo largo del tiempo')
ax.legend(loc='upper left')
plt.xticks(rotation=60)
plt.tight_layout()
plt.savefig('perdida_leads_tiempo.png')
plt.show()
```

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt

#  Variables independientes
Leads_Reales['pct_ilocalizable'] = Leads_Reales['Ilocalizable'] / Leads_Reales['Total'] * 100
Leads_Reales['pct_no_contesta']  = Leads_Reales['No Contesta']  / Leads_Reales['Total'] * 100
Leads_Reales['pct_incorrectos']  = Leads_Reales['D. Incorrectos'] / Leads_Reales['Total'] * 100
Leads_Reales['pct_duplicados']   = Leads_Reales['Duplicados']   / Leads_Reales['Total'] * 100

predictores = ['Total', 'pct_ilocalizable', 'pct_no_contesta',
               'pct_incorrectos', 'pct_duplicados', 'Abiertos', 'Año']
X = Leads_Reales[predictores].astype(float)
y = Leads_Reales['Conversion'].astype(float)

# Estandarizamos (z-score) para comparar el tamaño del efecto entre variables de escalas distintas ---
Xz = (X - X.mean()) / X.std()
yz = (y - y.mean()) / y.std()

modelo = sm.OLS(yz, sm.add_constant(Xz)).fit()
print(modelo.summary())
```

```python
# Gráfica de coeficientes
coefs = modelo.params.drop('const').sort_values()
pvals = modelo.pvalues.drop('const').reindex(coefs.index)
colores = ['#c44e52' if p < 0.05 else '#8c8c8c' for p in pvals]

fig, ax = plt.subplots(figsize=(8,5))
ax.barh(coefs.index, coefs.values, color=colores)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Coeficiente estandarizado')
ax.set_title('Efecto de cada variable sobre la Conversión (rojo = significativo, p<0.05)')
plt.tight_layout()
plt.savefig('regresion_conversion.png')
plt.show()
```
