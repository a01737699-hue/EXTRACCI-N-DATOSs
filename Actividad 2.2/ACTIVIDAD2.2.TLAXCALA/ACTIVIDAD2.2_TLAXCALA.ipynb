```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_excel("Base_de_Tlaxcala.xlsx")

data.head()
```

```python
data.shape
```

```python
valores_nulos = data.isnull().sum()
valores_nulos
```

```python
data.isnull().sum().sum()
```

```python
data2 = data.copy()
```

```python
# Eliminar columnas completamente vacías

data2 = data2.drop(["Telefono", "Correo"], axis=1)
```

```python
data2.columns
```

```python
data2.shape
```

Las variables Teléfono y Correo presentaron 56 valores nulos de un total de 56 registros, es decir, el 100% de sus datos se encontraba vacío. Debido a que no existe información disponible que permita realizar una sustitución confiable, se decidió eliminar ambas columnas. 

```python
data2["Producto"].isnull().sum()
```

```python
data["Producto"].mode()
```

```python
data["Producto"].value_counts(dropna=False)
```

```python
# Sustituir el valor nulo de Producto por un valor constante

data2["Producto"] = data2["Producto"].fillna(
    "Producto no registrado"
)
```

En la variable Producto se identificó 1 valor nulo. Aunque se trata de una variable categórica, se decidió no utilizar la moda, ya que asignar el producto más frecuente no garantiza que corresponda realmente al registro faltante. Por esta razón, el valor nulo fue sustituido por “Producto no registrado”, permitiendo conservar la observación sin generar información que no está disponible en la base de datos.

```python
# Corroborar valores nulos en todo el dataframe
valores_nulos_total = data2.isnull().sum().sum()
valores_nulos_total
```
