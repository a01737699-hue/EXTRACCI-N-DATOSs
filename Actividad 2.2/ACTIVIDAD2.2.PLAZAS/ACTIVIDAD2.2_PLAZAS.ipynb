```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = pd.read_excel("Plazas.xlsx")

data.head()
```

```python
data.shape
```

```python
valores_nulos = data.isnull().sum()
valores_nulos
```

```python
columnas_base = [
    "Fecha Origen",
    "Nombre",
    "Telefono",
    "Producto",
    "Asesor",
    "Origen"
]

data = data[data[columnas_base].notna().any(axis=1)].copy()

# Reiniciar índices
data = data.reset_index(drop=True)

data.shape
```

```python
data2 = data.copy()
```

```python
valores_nulos = data2.isnull().sum()
valores_nulos
```

# FECHA ORIGEN - 2 valores nulos 

```python
# Sustitución de valores nulos por un string en concreto

data2["Fecha Origen"] = data2["Fecha Origen"].astype("object")

data2["Fecha Origen"] = data2["Fecha Origen"].fillna(
    "Fecha no registrada"
)
```

En la variable Fecha  se encontraron 2 valores nulos. Se decidió utilizar el texto “Fecha no registrada”, ya que no existe información suficiente para determinar la fecha real de estos registros. Aunque podrían utilizarse métodos como ffill o bfill, estos asignarían la fecha de otro registro y podrían generar información incorrecta. También se consideró eliminar las filas, pero se prefirió conservarlas debido a que contienen información útil en las demás variables.

# NOMBRE - 1 VALOR NULO 

```python
# Eliminar fila que funciona como separador de mes
data2 = data2[
    ~((data2["Fecha Origen"] == "JUNIO") & (data2["Nombre"].isnull()))
].copy()

# Reiniciar índices
data2 = data2.reset_index(drop=True)
```

```python
data2.shape
```

Al analizar el valor nulo de la variable Nombre, se identificó que no correspondía a un cliente, sino a una fila utilizada como separador para indicar el inicio del mes de junio. Debido a que esta fila no representa una observación real dentro de la base de datos, se decidió eliminarla.

# TELEFONO

```python
# Eliminar columna Telefono porque está completamente vacía
data2 = data2.drop(["Telefono"], axis=1)
```

```python
valores_nulos = data2.isnull().sum()
valores_nulos
```

# SEGUIMIENTO 

```python
# Técnica: imputación por valor constante

data2["Seguimiento"] = data2["Seguimiento"].astype("object")

data2["Seguimiento"] = data2["Seguimiento"].fillna(
    "Seguimiento no registrado"
)
```

```python
valores_nulos = data2.isnull().sum()
valores_nulos
```

# CITA EFECTIVA

```python
# Técnica: imputación por valor constante

data2["Cita efetiva "] = data2["Cita efetiva "].astype("object")

data2["Cita efetiva "] = data2["Cita efetiva "].fillna(
    "Cita no registrada"
)
```

# PDM

```python
# Técnica: imputación por valor constante

data2["PDM"] = data2["PDM"].astype("object")

data2["PDM"] = data2["PDM"].fillna(
    "PDM no registrado"
)
```

#  SDC

```python
# Técnica: imputación por valor constante

data2["SDC"] = data2["SDC"].astype("object")

data2["SDC"] = data2["SDC"].fillna(
    "SDC no registrado"
)
```

```python
valores_nulos = data2.isnull().sum()
valores_nulos
```

# ESTATUS

```python
# Técnica: imputación por valor constante

data2["Estatus"] = data2["Estatus"].fillna(
    "Estatus no registrado"
)
```

```python
# Técnica: imputación por valor constante
# Aplicada a las preguntas abiertas restantes

columnas_preguntas = [
    "¿Por que no hay seguimiento?",
    "¿Por que no ha visitado la agencia?",
    "¿Por que no hubo prueba de manejo ?",
    "¿Por que no hubo solicitud de credito?",
    "¿Por que no se ha formalizado la compra?"
]

data2[columnas_preguntas] = data2[columnas_preguntas].fillna(
    "Sin informacion registrada"
)
```

```python
valores_nulos = data2.isnull().sum()
valores_nulos
```

```python
# Renombrar Unnamed: 16 porque contiene comentarios útiles
data2 = data2.rename(columns={
    "Unnamed: 16": "Comentarios adicionales"
})

# Integrar la información de Unnamed: 19 en Comentarios adicionales
data2["Comentarios adicionales"] = (
    data2["Comentarios adicionales"]
    .fillna(data2["Unnamed: 19"])
)

# Eliminar columnas innecesarias
data2 = data2.drop(
    ["Unnamed: 17", "Unnamed: 18", "Unnamed: 19"],
    axis=1
)

# Sustituir los comentarios que siguen vacíos
data2["Comentarios adicionales"] = (
    data2["Comentarios adicionales"]
    .fillna("Sin comentario adicional")
)
```
