```python
raw1 = pd.read_excel('Análisis_de_Piso.xlsx', header=None)
raw1.head(5)
```

Este reporte no tiene un encabezado simple de una fila: la fila 0 trae las categorías (Leads Piso 2024, Ventas Piso 2025, etc.), la fila 1 son totales por mes, y la fila 2 trae los nombres de los meses. Los datos de cada asesor empiezan hasta la fila 3. Por eso lo cargamos primero "en crudo" (header=None) para reconstruir la tabla correctamente.

```python
meses_leads2024 = ['feb','mar','abr','may','jun','jul','ago','sep','oct','nov','dic']
meses_leads2025 = ['ene','feb','mar','abr']
meses_ventas2024 = ['feb','mar','abr','may','jun','jul','ago','sep','oct','nov','dic']
meses_ventas2025 = ['ene','feb','mar','abr','may','junio']

columnas_piso = ['Asesor']
columnas_piso += [f'Leads_2024_{m}' for m in meses_leads2024]
columnas_piso += [f'Leads_2025_{m}' for m in meses_leads2025]
columnas_piso += ['Total_Leads','Total_Leads_Ajustado','Raking_Leads','Raking_Leads_Ajustado']
columnas_piso += [f'Ventas_2024_{m}' for m in meses_ventas2024]
columnas_piso += [f'Ventas_2025_{m}' for m in meses_ventas2025]
columnas_piso += ['Leads_Prom_X_Mes','T_Ventas_Piso','Factor','Total_Ventas_Historicas','Raking_Ventas',
                  'Dependencia_de_Piso','Historico_PDM','Historico_SDC','Conversion_PDM','Conversion_SDC',
                  'Factor_PDM','Factor_SDC','Raking_PDM','Raking_SDC']

len(columnas_piso)
```

Se arman los nombres de columna combinando la categoría (fila 0) con el mes correspondiente (fila 2), leyendo directamente la estructura del reporte de Excel. Esto reemplaza el encabezado automático de pandas, que no puede interpretar celdas combinadas.

```python
data1 = raw1.iloc[3:].copy()
data1.columns = columnas_piso
data1 = data1.reset_index(drop=True)
data1.head()
```

Se descartan las primeras 3 filas (categorías, totales y nombres de mes) porque no son datos de un asesor, y se asignan los nombres de columna reconstruidos en el chunk anterior.

```python
data1 = data1.replace(r'^\s*$', np.nan, regex=True)
```

Varias celdas "vacías" del Excel original en realidad tienen un espacio en blanco oculto (carácter \xa0) en vez de estar realmente vacías, así que pandas no las detecta como nulas con isnull(). Esta línea convierte cualquier celda que solo tenga espacios en blanco a un NaN real, para poder identificarlas y tratarlas correctamente.

```python
valores_nulos = data1.isnull().sum()
valores_nulos
```

```python
valores_nulos = data1.isnull().sum().sum()
valores_nulos
```

```python
data1_limpio = data1.copy()
```

```python
columnas_numericas = [c for c in data1_limpio.columns if c != 'Asesor']
data1_limpio[columnas_numericas] = data1_limpio[columnas_numericas].fillna(0)
```

Técnica: sustitución por un valor en concreto (0). Se aplicó a todas las columnas numéricas (leads mensuales, ventas mensuales, y las métricas calculadas como Factor, Conversión y Raking) porque, en este reporte, una celda vacía significa que ese asesor no tuvo actividad ese mes o no generó el indicador correspondiente — no es un dato que falte por error de captura, así que sustituirlo con media, mediana o ffill/bfill inventaría actividad que nunca ocurrió. La única columna sin tratar es "Asesor" (texto), que no tiene valores nulos.

```python
valores_nulos = data1_limpio.isnull().sum()
valores_nulos
```

```python
valores_nulos = data1_limpio.isnull().sum().sum()
valores_nulos
```

```python
data1_limpio.to_csv("analisis_piso_sin_nulos.csv", index=False)
```

```python
raw2 = pd.read_excel('Ventas_por_Asesor.xlsx', header=None)
raw2.head(5)
```

Igual que con Análisis de Piso, este reporte trae varias filas de encabezado antes de los datos reales: la fila 1 tiene las fechas de cada mes (2024-02-01, 2024-03-01, etc.) y la fila 2 tiene la etiqueta "Nombre vendedor" y algunas columnas de resumen (Raking, Acum, Ranking). Los datos de cada vendedor empiezan hasta la fila 3, y la última fila del archivo es "Total general" (un renglón de totales, no un vendedor real), así que también hay que excluirla.

```python
fechas = raw2.iloc[1, 1:31]
meses_cols = [f"Ventas_{d.strftime('%Y-%m')}" for d in fechas]

columnas_ventas = ['Nombre_Vendedor'] + meses_cols + ['Col_811', 'Raking',
                   'Acum_2024','Ranking_2024','Acum_2025','Ranking_2025','Acum_2026','Ranking_2026']

len(columnas_ventas)
```

A diferencia de Análisis de Piso (que usaba nombres de mes como texto), aquí la fila 1 ya trae fechas reales, así que se usan directamente para nombrar cada columna mensual (Ventas_2024-02, Ventas_2025-01, etc.). Las últimas 8 columnas son resumen: un valor sin etiqueta clara en el archivo original (Col_811), un raking general, y el acumulado/ranking de ventas por cada año (2024, 2025, 2026).

```python
data2 = raw2.iloc[3:54].copy()
data2.columns = columnas_ventas
data2 = data2.reset_index(drop=True)
data2.head()
```

Se toman solo las filas 3 a 53 (los 51 vendedores reales), dejando fuera las 3 filas de encabezado de arriba y la fila 54 ("Total general") que es un renglón de totales, no un vendedor.

```python
data2 = data2.replace(r'^\s*$', np.nan, regex=True)
```

```python
valores_nulos = data2.isnull().sum()
valores_nulos
```

```python
valores_nulos = data2.isnull().sum().sum()
valores_nulos
```

```python
data2_limpio = data2.copy()
```

```python
columnas_numericas2 = [c for c in data2_limpio.columns if c != 'Nombre_Vendedor']
data2_limpio[columnas_numericas2] = data2_limpio[columnas_numericas2].fillna(0)
```

Técnica: sustitución por un valor en concreto (0), igual razonamiento que en Análisis de Piso. Una celda vacía en las ventas mensuales significa que ese vendedor no vendió nada ese mes, y las columnas de resumen (Raking, Acum y Ranking por año) se derivan de esas mismas ventas, así que su ausencia también corresponde a 0 actividad. La única columna sin tratar es "Nombre_Vendedor" (texto), que no tiene valores nulos.

```python
valores_nulos = data2_limpio.isnull().sum()
valores_nulos
```

```python
valores_nulos = data2_limpio.isnull().sum().sum()
valores_nulos
```

```python
data2_limpio.to_csv("ventas_por_asesor_sin_nulos.csv", index=False)
```
