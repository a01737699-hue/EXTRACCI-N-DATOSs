```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

**VALORES NULOS**

```python
#Carga desde un archivo .csv sin indice
data = pd.read_csv('Citas_Digital_Filtrado.csv')
data.head(5)
```

```python
#Eliminar columnas innecesarias
#axis = 1 para columnas
#axis = 0 para filas
data2= data.drop(['Telefono ','Unnamed: 15'], axis=1)
data2
```

```python
#Identificar valores nulos por columna
valores_nulos=data2.isnull().sum()
valores_nulos
```

```python
#Identificar valores nulos por dataframe
valores_nulos=data2.isnull().sum().sum()
valores_nulos
```

**Métodos de Sustitución de Valores Nulos**

```python
#Realizamos una copia del dataframe
data3=data2.copy()
```

```python
#Primer método de sustitución de valores nulos
#Sustituir valores nulos con promedio o media
data3["PDM"]=data3["PDM"].fillna(round(data2["PDM"].mean(),1))
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Segundo método de sustitución de valores nulos
#Sustituir valores nulos con mediana
data3["SDC"]=data3["SDC"].fillna(round(data2["SDC"].median(),1))
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Tercer método de sustitución de valores nulos
#Sustituir valores nulos por un valor numérico en concreto
data3["Venta"]= data3["Venta"].fillna(0)
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Cuarto método de sustitución de valores nulos
#Sustituir valores nulos por un string en concreto
data3["Nombre Cliente "]= data3["Nombre Cliente "].fillna("Sin nombre registrado")
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un string en concreto
data3["Fecha"]= data3["Fecha"].fillna("Sin fecha registrada")
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un string en concreto
data3["Potencial de compra"]= data3["Potencial de compra"].fillna("Sin dato")
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un string en concreto
data3["¿Por que no ha visitado la agencia?"]= data3["¿Por que no ha visitado la agencia?"].fillna("No aplica")
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un string en concreto
data3["¿Por que no hay solicitud de credito?"]= data3["¿Por que no hay solicitud de credito?"].fillna("No aplica")
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un string en concreto
data3["¿Por que no hubo prueba de manejo ?"]= data3["¿Por que no hubo prueba de manejo ?"].fillna("No aplica")
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un string en concreto
data3["¿Por que no hubo proceso wow?"]= data3["¿Por que no hubo proceso wow?"].fillna("No aplica")
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por un string en concreto
data3["¿Por que se asigno antes de vistar piso?"]= data3["¿Por que se asigno antes de vistar piso?"].fillna("No aplica")
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Quinto método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia adelante "forward fill" ("ffill")
#Filtro por columnas
data3["BDC "] =data3["BDC "].ffill()
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sustituir valores nulos por valores no nulos hacia adelante forward fill (ffill)
#Filtro por columnas
data3["Asesor Asignado"] =data3["Asesor Asignado"].ffill()
```

```python
#Corroboramos valores nulos
valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Sexto método de sustitución de valores nulos
#Sustituir valores nulos por valores no nulos hacia atrás backward fill (bfill)
#Filtro por columnas
data3["Estatus de Lead"] =data3["Estatus de Lead"].bfill()
```

```python
#Corroboramos valores nulos en todo el dataframe
valores_nulos=data3.isnull().sum().sum()
valores_nulos
```
